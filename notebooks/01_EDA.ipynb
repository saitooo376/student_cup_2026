```python
c_cols = ['業界', '上場種別', '特徴']
c_cols_s = train.columns[train.columns.str.startswith("アンケート")].tolist()
c_cols = c_cols + c_cols_s
c_rows = len(c_cols)

fig, axes = plt.subplots(c_rows, 2, figsize=(10, 4*c_rows))

for i, col in enumerate(c_cols):

    train_counts = train[col].value_counts(dropna=True).sort_index()
    axes[i, 0].bar(train_counts.index.astype(str), train_counts.values, color="skyblue")
    axes[i, 0].set_title(f'train-{col}')

    test_counts = test[col].value_counts(dropna=True).sort_index()
    axes[i, 1].bar(test_counts.index.astype(str), test_counts.values, color="salmon")
    axes[i, 1].set_title(f'test-{col}')

    axes[i, 0].set_xticks(range(len(train_counts.index)))
    axes[i, 0].set_xticklabels(train_counts.index.astype(str), rotation=45)

    axes[i, 1].set_xticks(range(len(test_counts.index)))
    axes[i, 1].set_xticklabels(test_counts.index.astype(str), rotation=45)

plt.tight_layout()
plt.savefig("images/train_test_分布比較[カテゴリカル].png")
plt.close()
```

```python
n_cols = ['従業員数', '事業所数', '工場数','店舗数', '資本金', '総資産', '流動資産', '固定資産', 
          '負債', '短期借入金',	'長期借入金', '減価償却費'	]
n_rows = len(n_cols)

fig, axes = plt.subplots(n_rows, 2, figsize=(10, 4*n_rows))

for i, col in enumerate(n_cols):

    min_val = min(np.log1p(train[col].min()), np.log1p(test[col].min()))
    max_val = max(np.log1p(train[col].max()), np.log1p(test[col].max()))

    axes[i, 0].hist(np.log1p(train[col]), bins=30, color="skyblue")
    axes[i, 0].set_title(f'train-{col}')
    axes[i, 0].set_xlim(min_val, max_val)

    axes[i, 1].hist(np.log1p(test[col]), bins=30, color="salmon")
    axes[i, 1].set_title(f'test-{col}')
    axes[i, 1].set_xlim(min_val, max_val)

plt.tight_layout()
plt.savefig("images/train_test_分布比較[対数値].png")
plt.close()
```

```python
fig, axes = plt.subplots(len(c_cols), 1, figsize=(8, 6*len(n_cols)))

for i, col in enumerate(c_cols):
    ax = axes[i]
    cross_tab = (
        pd.crosstab(train[col], train['購入フラグ'], normalize='index') * 100
    )
    cross_tab.plot(
        kind='bar',
        stacked=True,
        ax=ax,
        color=['#e0e0e0', '#4caf50'],  # 未購入:グレー, 購入:グリーン
    )
    ax.set_title(f'{col}別構成比')
    ax.set_ylim(0, 100)

plt.tight_layout()
plt.savefig('images/カテゴリ変数ごと購入率.png')
plt.close()
```

```python
n_rows = len(n_cols)

fig, axes = plt.subplots(n_rows, 1, figsize=(10, 3.5 * n_rows))

for i, col in enumerate(n_cols):

  train_log = np.log1p(train[col])
  test_log = np.log1p(test[col])

  min_val = min(train_log.min(), test_log.min())
  max_val = max(train_log.max(), test_log.max())

  sns.histplot(
      data=train,
      x=train_log,
      hue='購入フラグ',
      bins=30,
      palette={0: 'skyblue', 1: 'orange'},  # 0:未購入, 1:購入 の色指定
      element='step',  # 輪郭線で重ねる見せ方 ('bars' や 'poly' も可能)
      stat='density',  # 割合（密度）で比較したい場合は有効化
      common_norm=False,  # グループごとの合計を1に正規化
      ax=axes[i],
  )
  axes[i].set_title(f'train-{col} 購入率)')
  axes[i].set_xlim(min_val, max_val)
  axes[i].set_xlabel(f'log1p({col})')

plt.tight_layout()
plt.savefig('images/数値データ購入率[対数値].png', dpi=300)
plt.close()
```

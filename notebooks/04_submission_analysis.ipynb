```python
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
```

```python
def save_confusion_matrix(exp_dir_name: str) -> None:
    """
    指定した実験ディレクトリのOOF予測値と閾値から混同行列を生成し、画像として保存する。
    """
    # メトリクスとOOF予測値の読み込み
    with open(f"../outputs/{exp_dir_name}/metrics.json", "r", encoding="utf-8") as f:
        metrics = json.load(f)

    oof = pd.read_csv(f"../outputs/{exp_dir_name}/oof.csv")

    # 正解ラベルと予測ラベルの生成
    y_true = oof["target"]
    pred_label = (oof["prediction"] > metrics["best_threshold"]).astype(int)

    # 混同行列の計算
    cm = confusion_matrix(y_true, pred_label)

    print(f'適合率: {cm[1,1] / sum(cm[:, 1])}')
    print(f'再現率: {cm[1,1] / sum(cm[1, :])}')

    # 描画処理
    plt.figure(figsize=(6, 5))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=['Predicted 0', 'Predicted 1'],
                yticklabels=['Actual 0', 'Actual 1'])

    plt.xlabel('Predicted')
    plt.ylabel('Actual')
    plt.title(f'Confusion Matrix ({exp_dir_name})')
    plt.tight_layout()
    
    # 保存とクローズ
    plt.savefig(f'images/{exp_dir_name}_confusion_matrix.png')
    plt.close()
```

### exp001

```python
save_confusion_matrix('exp001')
```

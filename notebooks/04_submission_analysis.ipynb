```python
import numpy as np
import pandas as pd
from sklearn.metrics import f1_score
```

### exp001

```python
oof = pd.read_csv("../outputs/exp001/oof.csv")

y_true = oof["target"]
oof_pred = oof["prediction"]
```

```python
best_f1 = 0
best_threshold = 0


for t in np.arange(0.05,0.95,0.05):

    pred_label = (oof_pred >= t).astype(int)

    score = f1_score(
        y_true,
        pred_label
    )

    if score > best_f1:
        best_f1 = score
        best_threshold = t


print(best_threshold)
print(best_f1)
```
